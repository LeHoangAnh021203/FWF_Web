**Bối cảnh:** FPT Telecom — 8 chi nhánh tỉnh và 20 giao dịch dịch vụ

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

## B1.1 — Tiền xử lý dữ liệu chi nhánh

```python
# TODO 1: đọc file fpt_chi_nhanh.csv vào biến df
df = pd.read_csv("data/fpt_chi_nhanh.csv")

# TODO 2: kiểm tra có giá trị thiếu (missing) ở cột nào không
missing_counts = df.isna().sum()
print(missing_counts)

# TODO 3: tạo cột mới "muc_do_rui_ro" = do_tre_gio * ty_le_loi
df["muc_do_rui_ro"] = df["do_tre_gio"] * df["ty_le_loi"]
print(df[["chi_nhanh", "muc_do_rui_ro"]].sort_values("muc_do_rui_ro", ascending=False))
```

## B1.2 — Phân cụm 8 chi nhánh bằng K-Means

```python
# TODO 4: chọn các cột số dùng để phân cụm
cot_so = ["do_tre_gio", "ty_le_loi", "yeu_cau_ho_tro", "hai_long_nld", "kpi_dat", "chi_phi_bi", "uptime"]
X = df[cot_so]

# TODO 5: chuẩn hoá dữ liệu bằng StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# TODO 6: chạy K-Means với K = 3 cụm, random_state=42
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["cum"] = kmeans.fit_predict(X_scaled)

print(df[["chi_nhanh", "cum"]])
print(df.groupby("cum")[cot_so].mean())
```

## B1.3 — Luật kết hợp dịch vụ bằng Apriori

```python
# TODO 7: đọc file fpt_dich_vu_giao_dich.csv vào biến giao_dich
giao_dich = pd.read_csv("data/fpt_dich_vu_giao_dich.csv")

# TODO 8: chuyển sang định dạng ma trận giỏ hàng bằng pandas
gio_hang = pd.crosstab(giao_dich["ma_giao_dich"], giao_dich["dich_vu"]) > 0
print(gio_hang.head())
print("Kích thước ma trận:", gio_hang.shape)

# TODO 9: cài mlxtend nếu chưa có (chạy một lần: !pip install mlxtend)
from mlxtend.frequent_patterns import apriori, association_rules

# TODO 10: chạy Apriori với support tối thiểu 20%
frequent_itemsets = apriori(gio_hang, min_support=0.2, use_colnames=True)

# TODO 11: tạo luật kết hợp với confidence tối thiểu 50%
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.5)
print(rules[["antecedents", "consequents", "support", "confidence", "lift"]])
```
